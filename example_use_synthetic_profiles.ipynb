```python
from synthetic_profiles import PVProfile, HydroProfile, DrinkingWaterProfile
```

```python
# Beispiel: PV-Profil für Terfens (2019–2023)
pv = PVProfile(
    lat=47.322360,
    lon=11.642956,
    peak_power_kw=5.0,
    loss=14.0,
    optimal_angles=True,
    startyear=2019,
    endyear=2023,
    name="Terfens PV",
    )

print(f'optimierte Neigung: {pv.angle:.1f}° und Ausrichtung: {pv.azimuth:.1f}°')
print(f'Jahreserzeugung: {pv.annual_energy_kwh():.1f} kWh pro Jahr')

# Profil abrufen
pv_erzeugung = pv.profile

# Skaliert auf andere Anlagengröße
pv_erzeugung_skaliert = pv.scaled(100.0)
```

```python
# Wasserkraftwerk mit 500 kW Leistung, Standard-Jahresprofile für 2019–2023
hydro = HydroProfile(
    startyear=2019,
    endyear=2023,
    installed_capacity_kw=500.0
    )

print(f'Wasserkraftwerk: {hydro.annual_energy_kwh():.1e} kWh pro Jahr')
```

```python
# Wasserkraftwerk mit 500 kW Leistung
# Eigene Monatsfaktoren (z.B. gemessene Durchflüsse normiert) für 2019–2023
hydro_custom = HydroProfile(
    startyear=2019,
    endyear=2023,
    installed_capacity_kw=500.0,
    monthly_factors=[0.2, 0.2, 0.4, 0.8, 1.0, 0.9, 0.7, 0.6, 0.5, 0.4, 0.3, 0.2],
    )

print(f'Wasserkraftwerk mit eigenen Monatsfaktoren: {hydro_custom.annual_energy_kwh():.1e} kWh pro Jahr')
```

```python
# Trinkwasserkraftwerk mit 15 kW Leistung
twkw = DrinkingWaterProfile(installed_capacity_kw=15.0) 

print(f'Trinkwasserkraftwerk: {twkw.annual_energy_kwh():.0f} kWh pro Jahr')
```
